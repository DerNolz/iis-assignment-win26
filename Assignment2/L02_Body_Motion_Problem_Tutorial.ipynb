## Answers:

## E1)

The robot achieved the nominal goal of placing the mug on the table, but it failed to satisfy the constraints of the entire action sequence.
The task requires success of the complete execution, not just a successful final mug position.


**Violated whole-course constraints:**

1. Spill prevention: Milk must be transferred without spilling onto the counter.

2. Environmental restoration: The refrigerator door must be closed after retrieving the milk carton.


## E2)

**Working Interpretation:** (:type serving) (:recipient Eva) (:substance milk)

**Inferred Function/Means:** (:vessel-role mug)

**Grounded Physical Object:** (:object Carton A)


## E3)

**Predicted effect:**

A low lift tests whether the grasp can support the carton's weight without slipping.

**Cost of failure:**

High carry: Failure will drop the carton, spill milk, damage objects, or contaminate the kitchen.

Low test: Low stakes, a few drops may be spilled, but almost nothing to nothing.

**Single piece of evidence:**

The asymmetric failure cost: The low test failure is relatively inexpensive, but high carry failure can have serious consequences.
The robot should test the grasp, detect slipping, adapt its grasp, and only then commit to high carry.


## E4)

The stopping rule depends on the properties of the poured material and the intended outcome.

1. **Milk into coffee**\
(:stop-when coffee-level reaches desired-level)\
Milk flows relatively easily, so the robot should monitor the liquid level and stop pouring when the coffee reaches the desired level.

2. **Pancake batter into a pan**\
(:stop-when disc-size reaches 90% of target-size)\
Pancake batter flows more slowly than milk. It continues flowing outwards after the pour was stopped. (90% is an approximation, may be different)

**Difference in wrist control:**

With milk, the robot can regulate the pour quickly by monitoring the rising liquid level and tilting the wrist back near the target.

With batter, the robot must anticipate delayed flow and begin tilting back earlier to compensate for residual material leaving the container and flowing outwards.

This alters the material dynamics and the timing of the stopping command.


## E5)

1. Infer\
Understand that Eva wants milk in her coffee.

2. Ground\
Identify Carton A as the milk source.

3. Predict\
Predict whether the selected grasp can support the carton.

4. Decide\
Select low (5 mm) test lift instead of high carry.

5. Regulate\
Execute the small lift while monitoring movement.

6. Detect & Revise\
Detect downward motion, identify slipping, revise the grasp.

7. Verify\
Confirm stable grasping and successful completion without spills.

**Which step was triggered when downward motion was observed during the 5 mm lift?**

Detect & Revise. (Stage 6)\
The robot observed unexpected downward motion during the 5 mm lift, which indicated that the original grasp was unstable.\
This leads to revision of the action designator by introducing a grasp adaptation.

**How was the action designator developed in response?**

A grasp adaptation takes place by recalculation forces and grip.


## E6)

**Failed ablation system:**

- The milk carton has a wet, slippery surface.
- The carton is positioned on a narrow refrigerator shelf.
- The carton is relatively heavy because it is nearly full.

The high-side-pinch grasp may initially appear successful but slip during lifting. It might carry the unstable carton away from the shelf, drop it and spill milk.

**Why does single-run execution success differ from evidence of general competence across task variants?**

A single successful execution only demonstrates the robot succeeding under one particular set of conditions.\
General competence requires successful adaptation across different task variants, e.g:\
Dry and wet carton surfaces, different carton weights, different shelf positions, different object geometries, different liquid materials, etc...
